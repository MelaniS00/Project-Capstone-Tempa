```python
# 3. Fungsi untuk membersihkan format c("bahan1", "bahan2") menjadi list Python seutuhnya
def bersihkan_bahan(teks_bahan):
    try:
        # Menghapus pola c() bawaan dari CSV
        teks_bersih = re.sub(r'^c\((.*)\)$', r'[\1]', teks_bahan)
        # Mengubah string '["bahan1", "bahan2"]' menjadi list asli ['bahan1', 'bahan2']
        return ast.literal_eval(teks_bersih)
    except:
        return []
```

```python
# 1. Buat fungsi pembersih khusus format c("...", "...")
def bersihkan_format_r(teks):
    # Jika data kosong (NaN)
    if pd.isna(teks):
        return []

    # Jika data ternyata sudah berbentuk list murni
    if isinstance(teks, list):
        return teks

    # Jika data berbentuk string, lakukan pembersihan
    if isinstance(teks, str):
        teks = teks.strip()
        # Mendeteksi jika string diawali dengan 'c(' dan diakhiri ')'
        if teks.startswith("c(") and teks.endswith(")"):
            # Ubah c("bahan1", "bahan2") menjadi ["bahan1", "bahan2"]
            teks = "[" + teks[2:-1] + "]"

        try:
            # Ubah string teks tadi menjadi List Python sesungguhnya
            return ast.literal_eval(teks)
        except:
            return [] # Jika gagal parse, kembalikan list kosong

    return []
```

```python
# 1. Bikin fungsi pembersih baru berbasis REGEX
def bersihkan_bahan_regex(teks):
    if pd.isna(teks):
        return []

    if isinstance(teks, list):
        return teks

    if isinstance(teks, str):
        # Mengambil semua teks yang diapit oleh tanda kutip ganda "..." atau tunggal '...'
        # Cara ini 100% aman dari error malformed string
        ekstrak_bahan = re.findall(r'["\']([^"\']+)["\']', teks)

        # Ubah langsung ke huruf kecil semua (lowercase) agar seragam
        return [bahan.lower().strip() for bahan in ekstrak_bahan]

    return []

# 2. Terapkan fungsi baru ini ke dataset kamu
df_clean['RecipeIngredientParts'] = df_clean['RecipeIngredientParts'].apply(bersihkan_bahan_regex)

# 3. KITA CEK LANGSUNG HASILNYA
print("--- CEK HASIL PEMBERSIHAN ---")
print("Baris pertama sekarang berisi:")
print(df_clean['RecipeIngredientParts'].iloc[0])
```

```python
# =====================================================================
# 1. FUNGSI PEMBERSIH KOLOM RecipeInstructions
# =====================================================================
def bersihkan_instruksi(teks):
    if pd.isna(teks):
        return ""
    if not isinstance(teks, str):
        return str(teks)

    # Ekstrak semua teks yang berada di dalam tanda kutip (langkah-langkah memasak)
    langkah_langkah = re.findall(r'["\']([^"\']+)["\']', teks)

    if not langkah_langkah:
        # Jika format aneh/bukan c(...), bersihkan karakter dasar saja
        teks_bersih = re.sub(r'^c\(|\)$', '', teks)
        return teks_bersih.replace('""', '"').strip()

    langkah_bersih = []
    for langkah in langkah_langkah:
        # A. Hilangkan enter tersembunyi (\n atau \r) dan tab (\t) di dalam teks langkah
        langkah = re.sub(r'[\r\n\t]+', ' ', langkah)
        # B. Hilangkan spasi ganda/aneh yang berlebihan menjadi satu spasi normal
        langkah = re.sub(r'\s+', ' ', langkah)
        langkah = langkah.strip()

        if langkah:
            langkah_bersih.append(langkah)

    # C. Gabungkan menjadi teks berurutan (1. ..., 2. ...) dipisahkan enter normal (\n)
    # Format ini otomatis dibaca sangat rapi oleh Streamlit markdown
    instruksi_final = "\n".join([f"{i+1}. {l}" for i, l in enumerate(langkah_bersih)])
    return instruksi_final

# Terapkan ke kolom instruksi
df_clean['RecipeInstructions'] = df_clean['RecipeInstructions'].apply(bersihkan_instruksi)
```

```python
# =====================================================================
# 2. FUNGSI PEMBERSIH KOLOM Images (SUDAH DIPERBAIKI)
# =====================================================================
def ambil_satu_gambar(teks):
    if pd.isna(teks):
        return ""
    if not isinstance(teks, str):
        return ""

    # PERBAIKAN: Menghapus koma (,) dari list pengecualian Regex.
    # Sekarang diperbolehkan membaca koma agar parameter gambar (w_555,h_416) tidak terpotong.
    # Regex ini akan berhenti tepat saat menyentuh tanda kutip (" atau ') atau tanda kurung tutup ).
    urls = re.findall(r'https?://[^"\s\')]+', teks)

    if urls:
        # Ambil hanya elemen pertama (link gambar pertama dari baris tersebut)
        link_pertama = urls[0].strip()
        return link_pertama

    return "" # Kembalikan string kosong jika resep tidak punya gambar

# =====================================================================
# JALANKAN ULANG PADA DATASET KAMU
# =====================================================================
df_clean['Images'] = df_clean['Images'].apply(ambil_satu_gambar)
```

# Model

```python
class RecipeRecommender:
    def __init__(self, df_recipes):
        self.df = df_recipes
        self.commodity_mapping = {
            'meat_beef': 'beef',
            'meat_chicken': 'chicken',
            'meat_chicken_broiler': 'chicken',
            'eggs': 'egg',
            'onions': 'onion',
            'chili': 'chili',
            'garlic': 'garlic',
            'oil': 'oil',
            'rice': 'rice',
            'sugar': 'sugar'
        }

    def _normalize_input(self, user_inputs):
        normalized = [self.commodity_mapping.get(item, item) for item in user_inputs]
        return set(normalized)

    def recommend(self, user_inputs):
        input_set = self._normalize_input(user_inputs)
        df_model = self.df.copy()

        # =========================================================================
        # PERBAIKAN: Ditambahkan 'if isinstance(x, list) else []'
        # Jika baris data berupa NaN (float), sistem otomatis mengabaikannya (dianggap list kosong)
        # =========================================================================

        # Tahap 1: Filtering (Mencari irisan)
        df_model['matched_ingredients'] = df_model['RecipeIngredientParts'].apply(
            lambda x: list(input_set.intersection(set(x))) if isinstance(x, list) else []
        )
        df_model['match_count'] = df_model['matched_ingredients'].apply(len)

        # Ambil yang minimal cocok 1 bahan
        candidates = df_model[df_model['match_count'] > 0].copy()

        # Tahap 2: Scoring (Menghitung bahan yang kurang)
        candidates['missing_count'] = candidates['RecipeIngredientParts'].apply(
            lambda x: len(set(x) - input_set) if isinstance(x, list) else 0
        )
        candidates['missing_ingredients'] = candidates['RecipeIngredientParts'].apply(
            lambda x: list(set(x) - input_set) if isinstance(x, list) else []
        )

        # Tahap 3: Ranking
        final_recommendations = candidates.sort_values(by=['match_count', 'missing_count'], ascending=[False, True])
        return final_recommendations

# ==========================================
# 2. SIMULASI DATA (Disesuaikan dengan selected_cols kamu)
# ==========================================
data_dummy = {
    'RecipeId': [1, 2, 3, 4, 5],
    'Name': ['Nasi Goreng Spesial', 'Telur Ceplok Bawang', 'Ayam Pedas Manis', 'Tumis Bawang Putih', 'Sup Ayam Bening'],
    'RecipeIngredientParts': [
        ['rice', 'egg', 'onion', 'garlic', 'chili', 'oil', 'salt'],
        ['egg', 'onion', 'garlic', 'oil'],
        ['chicken', 'chili', 'sugar', 'onion', 'oil'],
        ['garlic', 'oil', 'salt'],
        ['chicken', 'garlic', 'salt', 'water']
    ],
    'RecipeInstructions': ['Langkah 1...', 'Langkah 1...', 'Langkah 1...', 'Langkah 1...', 'Langkah 1...'],
    'Images': ['url1', 'url2', 'url3', 'url4', 'url5']
}
df_resep = pd.DataFrame(data_dummy)

# Inisialisasi Model untuk pengujian dummy
model = RecipeRecommender(df_resep)

# ==========================================
# 3. PENGUJIAN (TESTING SCENARIOS)
# ==========================================

def jalankan_tes(skenario_nama, input_user):
    print(f"--- MENGUJI: {skenario_nama} ---")
    print(f"Input User: {input_user}")

    hasil = model.recommend(input_user)

    if hasil.empty:
        print("-> Tidak ada resep ditemukan.\n")
        return

    for index, row in hasil.head(3).iterrows(): # Tampilkan top 3 saja
        print(f"Resep: {row['Name']}")
        print(f"   [+] Bahan Cocok ({row['match_count']}): {row['matched_ingredients']}")
        print(f"   [-] Kurang Bahan ({row['missing_count']}): {row['missing_ingredients']}\n")
    print("==========================================\n")

# Skenario 1: Input 1 Komoditi
jalankan_tes("Skenario 1 (Satu Bahan)", ['eggs'])

# Skenario 2: Input 2 Komoditi
jalankan_tes("Skenario 2 (Dua Bahan)", ['eggs', 'onions'])

# Skenario 3: Input 3 Komoditi
jalankan_tes("Skenario 3 (Tiga Bahan)", ['meat_chicken', 'chili', 'sugar'])
```

```python
# 1. Inisialisasi model menggunakan data asli df_clean
model_asli = RecipeRecommender(df_clean)

# 2. Masukkan komoditi yang dicari
input_kamu = ['eggs', 'chili', 'onions']

# 3. Panggil fungsi rekomendasi (Sekarang sudah aman dari error float!)
df_hasil_rekomendasi = model_asli.recommend(input_kamu)

# 4. Tampilkan hasil 5 teratas
df_hasil_rekomendasi.head()
```
